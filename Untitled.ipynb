```python
mean_age=df["Age"].mean()
df["Age"]=df["Age"].fillna(mean_age)
df
```

```python
df["Gender"]=df["Gender"].fillna("unknown")
```

```python
df.isnull().sum()
```

```python
df["Course"]=df["Course"].fillna("unknown") # couse are not mentioned
df["Subject"]=df["Subject"].fillna("unknown")#subject data are missing
df["Email"]=df["Email"].fillna("unknown")#email not provided
df["City"]=df["City"].fillna("unknown")#city dat is missing 
```

```python
df.isnull().sum()
```

```python
df['Course'] = df['Course'].str.lower()
df['Gender']=df["Gender"].str.lower()
df['City']=df["City"].str.lower()
```
